# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

### Creating Tuples

```python
# Empty
t1 = ()
print(t1)

# Create a tuple with a single item
t2 = ('hello')
print(t2)
print(type(t2))

# homo
t3 = (1,2,3,4)
print(t3)

# Hetro
t4 = (1,2,3,4,True,[1,2,3,4])
print(t4)

# Tuple
t5 = (1,2,3,(45,6))
print(t5)

# Using type conversion
t6 = tuple('Hello')
print(t6)
```

```python
import time

L = list(range(10000))
T = tuple(range(10000))

start = time.time()
for i in L:
    i*5
print('List time: ',time.time()-start)

start = time.time()
for i in T:
    i*5
print('Tuple time: ',time.time()-start)
```

```python
import sys

L = list(range(10000))
T = tuple(range(10000))

print('List size', sys.getsizeof(L))
print('Tuple size', sys.getsizeof(T))
```

# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

## Creating Sets

```python
# Empty
s = set()
print(s)
print(type(s))

# 1D and 2D 
s1 = {1,3,5}
print(s1)

# s2 = {1,2,3,{4,5}}
# print(s2)
# Hemo and hetro
s3 = {1,'hello',6.5,(1,2,3)}
print(s3)

# Using type conversion
s4 = set([1,2,3,4])
print(s4)

# # duplicates not allowed
# s5 = {1,1,2,2,3,3}
# print(s5)
# # set can't have mutable items
# s6 = {1,2,[3,4]}
# print(s6)
```

```python
s1 = {1,2,3}
s2 = {3,1,2}

print(s1 == s2)
```

### Adding Items

```python
s = {1,2,3,4}
# add
s.add(5)
print(s)

p = {4,5,6}
p.update([34,56])
print(p)
```

## Set Operation

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1 | s2
# Union(|)
# Intersection(&)
s1 & s2
# Difference(-)
s1 - s2
s2 - s1
# Symmetric Difference(^)
s1 ^ s2
# Membership Test
1 not in s1
# Iteration
for i in s1:
  print(i)
```

### Set Function

```python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse=True)
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1 | s2
s1.union(s1)

s1.update(s2)
print(s1)
print(s2)
```

```python
# intersection/intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)

s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)

s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2 = {7,8,5,6}

s1.isdisjoint(s2)
```

```python
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2)
```

```python
# copy
s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
```

# Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'Aman' , 'age' : 20 , 'gender' : 'male' }

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

## What are Functions?
* ###  Functions in Python are reusable blocks of code that perform a specific task. They help organize code, improve readability, and reduce redundancy. Python functions are created using the def keyword.

### Let's create a function(with docstring)

```python
def is_even(num):   # num is a 'parameter' here
    """
    This function returns if a given number is even or odd
    input - any valid integer
    output - odd/even
    """
    if type(num) == int:
        if num % 2 == 0:
            return 'even'
    
        else:
            return 'odd'
    else:
        return 'Try different input'


# fun if we create : *Parameter*
# fun if we use : *Argument*
```

```python
# Call the function
# function_name(input)
for i in range(1,11):
    x = is_even(i)
    print(x)
```

# *args and **kwargs
*args and **kwargs are special Python keywords that used to pass the variable length of arguments of a function.

```python
# *args 
# allows us to pass a variable number of non-keywords to a function
def multiply(*args):
    product = 1

    for i in args:
        product = product * i

    return product
    
```

```python
multiply(1,2,3,4,5,6)
```

```python
print(is_even.__doc__)  # Used to see the documentation
```

```python
# **Kwargs
# it allows us to pass any number of keyword arguments
# Keyword arguments mean that they contain a key-value pair, like a python dictionary.

def display(**kwargs):
    for (key,value) in kwargs.items():
        print(key,'->',value)
    
```

```python
display(india='Delhi',srilanka='colombo',nepal='katmandu')
```

##### Points to remember while using `*args and **kwargs`

- order of the arguments matter(normal -> `*args` -> `**kwargs`)
- The words “args” and “kwargs” are only a convention, you can use any name of your choice
